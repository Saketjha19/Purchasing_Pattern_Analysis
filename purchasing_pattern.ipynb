```python
import pandas as pd
```

```python
df=pd.read_csv("customer_shopping_behavior.csv")
```

```python
df.isnull().sum()
```

```python
# 37 null values that to be replaced
df["Category"].unique()
```

```python
# since it was different types of categories we can't just replca missing values by median of overall categories
# we need to replce the missing values by median of their respective categories
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
# changing the case for better reading and utilization in sql
```

```python
df.columns=df.columns.str.lower()
```

```python
df.columns=df.columns.str.replace(' ','_')
df.columns
```

```python
df.head(10)
```

```python
# Visualizing the age group distribution for the cusotmers

df["age_group"]=pd.qcut(df["age"],q=4,labels=["Young adult","Adult","Middle-aged","Senior"])
```

```python
df[["age","age_group"]]
```

```python
df.frequency_of_purchases
```

```python
# mapping
df["purchase_frequency_days"]=df["frequency_of_purchases"].map({
    'Fortnightly':14,'Weekly':7, 'Annually':365, 'Quarterly':90, 'Bi-Weekly':14,
       'Monthly':30, 'Every 3 Months':90
})
```

```python
df.frequency_of_purchases.unique()
```

```python
(df["discount_applied"]==df["promo_code_used"]).all()
```

```python
df[df["discount_applied"]!=df["promo_code_used"]]
# so actually both have the same result because when promo_code is applied then discount will defenitely prevail
```

```python
# so we can drop one coulmn
# df=df.drop('promo_code_applied',axis=1)
df.columns
```

```python
df=df.drop('promo_code_used',axis=1)
```

```python
df.head(3)
```

```python
# classifying customers into spending capacities
df["one_time_spending_capacity"]=pd.qcut(df["purchase_amount_(usd)"],q=3,labels=["High_spendor","Medium_spendor","Low_spendor"])
```
